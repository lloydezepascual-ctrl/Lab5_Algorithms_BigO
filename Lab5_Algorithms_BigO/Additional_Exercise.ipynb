```python
# --- Exercise 1: Emergency Dispatch Record Search ---

# 1. Student Inputs & Calculated Rules
LAST_NAME = "Pascual"
STUDENT_ID = "TUPM-26-4321"
FAVORITE_ARTIST = "ARTHUR NERY"

SEED_DIGIT = 1
ID_SUM = 18
NAME_LENGTH = len(LAST_NAME)          # 7
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 11
ARTIST_SUM = 148

# 2. Data Generation
base_dataset = [SEED_DIGIT + (i * NAME_LENGTH) for i in range(10)]
rotation_point = SEED_DIGIT % len(base_dataset)

# Rotated Dataset
rotated_dataset = base_dataset[rotation_point:] + base_dataset[:rotation_point]

# Target Incident (ID_SUM + ARTIST_SUM)
target_incident = ID_SUM + ARTIST_SUM  # 166

# 3. Linear Search Function
def linear_search(arr, target):
    comparisons = 0
    for i, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

# 4. Rotated Binary Search Function
def rotated_binary_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, comparisons
        
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

# 5. Run and Print Results
idx_lin, comp_lin = linear_search(rotated_dataset, target_incident)
idx_rot, comp_rot = rotated_binary_search(rotated_dataset, target_incident)

print("=== EXERCISE 1 ASSESSMENT DATA ===")
print(f"Original Dataset: {base_dataset}")
print(f"Rotated Dataset:  {rotated_dataset}")
print(f"Target Incident:  {target_incident}")
print(f"Linear Search -> Index Found: {idx_lin}, Comparisons: {comp_lin}")
print(f"Rotated Binary Search -> Index Found: {idx_rot}, Comparisons: {comp_rot}")
```

```python
# --- Exercise 2: University Enrollment Records ---

# 1. Student Inputs & Calculated Rules
LAST_NAME = "Pascual"
STUDENT_ID = "TUPM-26-4321"
FAVORITE_ARTIST = "ARTHUR NERY"

SEED_DIGIT = 1
ID_SUM = 18
NAME_LENGTH = len(LAST_NAME)          # 7
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 11
ARTIST_SUM = 148

# 2. Data Generation
data_unsorted = [(SEED_DIGIT * i + ID_SUM) % (ARTIST_SUM + 1) for i in range(1, 11)]

# 3. Bubble Sort Function
def bubble_sort(arr):
    arr_copy = arr.copy()
    n = len(arr_copy)
    swaps = 0
    comparisons = 0
    
    for i in range(n):
        for j in range(0, n - i - 1):
            comparisons += 1
            if arr_copy[j] > arr_copy[j + 1]:
                arr_copy[j], arr_copy[j + 1] = arr_copy[j + 1], arr_copy[j]
                swaps += 1
                
    return arr_copy, comparisons, swaps

# 4. Selection Sort Function
def selection_sort(arr):
    arr_copy = arr.copy()
    n = len(arr_copy)
    swaps = 0
    comparisons = 0
    
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr_copy[j] < arr_copy[min_idx]:
                min_idx = j
        if min_idx != i:
            arr_copy[i], arr_copy[min_idx] = arr_copy[min_idx], arr_copy[i]
            swaps += 1
            
    return arr_copy, comparisons, swaps

# 5. Run and Print Results
sorted_bubble, comp_b, swaps_b = bubble_sort(data_unsorted)
sorted_select, comp_s, swaps_s = selection_sort(data_unsorted)

print("=== EXERCISE 2 ASSESSMENT DATA ===")
print(f"Unsorted Dataset: {data_unsorted}")
print(f"Sorted Dataset:   {sorted_bubble}")
print(f"Bubble Sort    -> Comparisons: {comp_b}, Swaps: {swaps_b}")
print(f"Selection Sort -> Comparisons: {comp_s}, Swaps: {swaps_s}")
```

```python
# --- Exercise 3: E-Commerce Order Processing ---

# 1. Student Inputs & Calculated Rules
LAST_NAME = "Pascual"
STUDENT_ID = "TUPM-26-4321"
FAVORITE_ARTIST = "ARTHUR NERY"

SEED_DIGIT = 1
ID_SUM = 18
NAME_LENGTH = len(LAST_NAME)          # 7
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 11
ARTIST_SUM = 148

# 2. Data Generation
data_ex3 = [(ID_SUM + i * ARTIST_LENGTH) % 100 for i in range(10)]

# 3. Insertion Sort Function
def insertion_sort(arr):
    arr_copy = arr.copy()
    comparisons = 0
    swaps = 0
    for i in range(1, len(arr_copy)):
        key = arr_copy[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if arr_copy[j] > key:
                arr_copy[j + 1] = arr_copy[j]
                swaps += 1
                j -= 1
            else:
                break
        arr_copy[j + 1] = key
    return arr_copy, comparisons, swaps

# 4. Run and Print Results
sorted_insertion, comp_i, swaps_i = insertion_sort(data_ex3)

print("=== EXERCISE 3 ASSESSMENT DATA ===")
print(f"Dataset:        {data_ex3}")
print(f"Sorted Dataset: {sorted_insertion}")
print(f"Insertion Sort -> Comparisons: {comp_i}, Swaps: {swaps_i}")
```
